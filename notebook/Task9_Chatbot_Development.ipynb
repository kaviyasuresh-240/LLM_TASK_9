# Task 9 - Chatbot Development

## Building an Interactive AI Chatbot using Ollama

## Objective

- Understand the architecture of a local LLM chatbot.
- Build a command-line chatbot using Ollama and Gemma.
- Maintain conversation history.
- Store chat logs for future integration.

## Technologies Used

- Python
- Ollama
- Gemma
- Pandas
- CSV Files

## Observations

- The chatbot successfully answered healthcare-related questions.
- Context awareness was maintained through conversation history.
- Responses were stored automatically.
- Logs included timestamps.

## Strengths

- Runs locally without internet.
- Maintains conversation context.
- Generates detailed healthcare responses.
- Automatically stores conversations.

## Limitations

- May generate inaccurate medical advice.
- Response quality depends on prompt wording.
- Slower on low-end hardware.
- Not a substitute for professional healthcare consultation.

## Future Work

The current chatbot successfully performs healthcare-related conversations using the Gemma LLM through Ollama. However, several improvements can be implemented in future versions:

1. Integrate the chatbot with LangChain to enable advanced workflow management and prompt engineering.
2. Add Retrieval-Augmented Generation (RAG) using a healthcare knowledge base for more accurate responses.
3. Store conversation history in a database instead of CSV files for better scalability.
4. Develop a graphical user interface (GUI) using Streamlit or Gradio.
5. Implement speech-to-text and text-to-speech capabilities for voice-based interaction.
6. Add user authentication and personalized conversation histories.
7. Improve response accuracy through domain-specific healthcare datasets and fine-tuned models.
8. Deploy the chatbot as a web application for wider accessibility.

## Conclusion

An interactive healthcare chatbot was successfully developed using Python, Ollama, and the Gemma language model. The chatbot accepts continuous user input, generates contextual responses, maintains conversation history, and stores logs with timestamps. The project demonstrates the practical application of local Large Language Models for conversational AI and provides a foundation for future integration with LangChain and advanced AI workflows.
